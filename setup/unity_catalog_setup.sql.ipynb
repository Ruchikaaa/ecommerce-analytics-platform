## Domain 1: Unity Catalog Administration & Governance
This section initializes the 3-tier namespace (`catalog.schema.table`). By creating environment-specific catalogs, we ensure absolute isolation between Development and Production data.

```python
%sql
-- Create the environment-specific catalog for development
CREATE CATALOG IF NOT EXISTS ecommerce_analytics_dev
COMMENT 'Main catalog for e-commerce analytics platform - Development ';

USE CATALOG ecommerce_analytics_dev;

-- Create schemas for the Medallion layers
CREATE SCHEMA IF NOT EXISTS bronze_layer
COMMENT 'Schema for raw ingested data with audit columns';

CREATE SCHEMA IF NOT EXISTS silver_layer
COMMENT 'Schema for cleaned, deduplicated, and validated data';

CREATE SCHEMA IF NOT EXISTS gold_layer
COMMENT 'Schema for aggregated business metrics and dimensional models';
```

## Domain 1: Storage Governance & External Volumes
We define Governed Volumes to manage our files. External Volumes link our landing zone in ADLS Gen2, while Managed Volumes provide a secure home for DLT checkpoints.

```python
%sql
-- Define the External Volume for the ADLS Gen2 landing zone
-- This allows governed access to your 110M records across Oct/Nov CSV files
CREATE EXTERNAL VOLUME IF NOT EXISTS ecommerce_analytics_dev.bronze_layer.landing_csvs
LOCATION 'abfss://bronze@struchikaecommerce.dfs.core.windows.net/landing/'
COMMENT 'Governed landing zone for raw CSV files';

-- Define a Managed Volume for DLT Checkpoints
-- This replaces legacy DBFS with UC-governed storage for pipeline state management
CREATE VOLUME IF NOT EXISTS ecommerce_analytics_dev.bronze_layer.checkpoints
COMMENT 'Governed storage for DLT streaming checkpoints and schema metadata';
```
