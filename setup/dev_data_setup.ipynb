```python
# 1. Define Paths
source_path = "/Volumes/ecommerce_analytics_dev/bronze_layer/landing_csvs/"
dev_sample_path = "/Volumes/ecommerce_analytics_dev/bronze_layer/landing_csvs/dev/"

# 2. Create the sample with Schema Inference
print("Reading source with header to infer schema...")
df_sample = (spark.read.format("csv")
    .option("header", "true")        # Critical: Tells Spark to read column names from the first line
    .option("inferSchema", "true")   # Critical: Tells Spark to guess types (int, string, etc.)
    .load(source_path)
    .limit(100000))                  # Taking only 100k rows for Dev

# 3. Save to the new subfolder
print(f"Saving sample to: {dev_sample_path}")
(df_sample.write.format("csv")
    .option("header", "true")
    .mode("overwrite")
    .save(dev_sample_path))

print("Success! Sample created.")
```
